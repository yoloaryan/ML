 Basic Machine Learning

Now we'll build a Placement Prediction Model.

Problem

Predict whether a student will be placed based on attendance and academic performance.

First convert placement status into numbers:

```python
df["Placement_Target"] = df[
    "Placement_Status"
].map({
    "Placed": 1,
    "Not Placed": 0
})
```

```python
df[
    ["Placement_Status", "Placement_Target"]
].head()
```

Select Features

We'll use:

Attendance
Assignment Score
Midterm Score
Final Score
Overall Marks

```python
X = df[
    [
        "Attendance",
        "Assignment_Score",
        "Midterm_Score",
        "Final_Score",
        "Overall_Marks"
    ]
]

y = df["Placement_Target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
model = LogisticRegression()

model.fit(
    X_train,
    y_train
)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred)
```

Model Accuracy

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    "Model Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(cm)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.colorbar()

plt.show()
```
